```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
DATA_PATH = "../data/raw/retail_clean_dataset/"
ANOMALY_PATH = "../data/raw/retail_contaminated_dataset/"

customers = pd.read_csv(DATA_PATH + "customer_master.csv")

inventory = pd.read_csv(DATA_PATH + "inventory_snapshot.csv")

promotions = pd.read_csv(DATA_PATH + "promotions.csv")

sales = pd.read_csv(DATA_PATH + "sales_transactions.csv")

products = pd.read_csv(DATA_PATH + "sku_master.csv")

stores = pd.read_csv(DATA_PATH + "store_master.csv")

flags = pd.read_csv(
    ANOMALY_PATH + "sku_inventory_flags.csv"
)
```

```python
customers_clean = customers.copy()

inventory_clean = inventory.copy()

promotions_clean = promotions.copy()

sales_clean = sales.copy()

products_clean = products.copy()

stores_clean = stores.copy()

flags_clean = flags.copy()
```

```python
before = sales_clean.memory_usage(
    deep=True
).sum() / 1024**2

print(f"Before: {before:.2f} MB")
```

```python
sales_clean["channel"] = sales_clean[
    "channel"
].astype("category")

sales_clean["store_id"] = sales_clean[
    "store_id"
].astype("category")

sales_clean["promo_id"] = sales_clean[
    "promo_id"
].astype("category")

sales_clean["sku_id"] = sales_clean[
    "sku_id"
].astype("category")

sales_clean["customer_id"] = sales_clean[
    "customer_id"
].astype("category")
```

```python
sales_clean["quantity"] = pd.to_numeric(

    sales_clean["quantity"],

    downcast="integer"
)
```

```python
float_cols = [

    "unit_price",

    "total_value",

    "discount_pct"

]

for col in float_cols:

    sales_clean[col] = pd.to_numeric(

        sales_clean[col],

        downcast="float"
    )
```

```python
after = sales_clean.memory_usage(
    deep=True
).sum() / 1024**2

print(f"After: {after:.2f} MB")

saved = before - after

print(f"Saved: {saved:.2f} MB")
```
